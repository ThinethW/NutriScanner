```python
# ============================================================
# COMPLETE 3-MODEL SMART ENSEMBLE WITH TESTING
# ============================================================

# Install ultralytics if not already installed
!pip install ultralytics --quiet

from google.colab import drive
drive.mount('/content/drive')

from ultralytics import YOLO
import os
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# ============================================================
# MODEL PATHS
# ============================================================

model_paths = {
    'V21': "/content/drive/MyDrive/TeamProject/models/srilankan_food_model_v21_74.5.pt",
    'V24': "/content/drive/MyDrive/TeamProject/models/srilankan_food_model_v24_71.9.pt",
    'V25': "/content/drive/MyDrive/TeamProject/models/srilankan_food_model_v25_70.5.pt"
}

# ============================================================
# 3-MODEL SMART ENSEMBLE CLASS
# ============================================================

class ThreeModelEnsemble:
    def __init__(self, v21_path, v24_path, v25_path):
        print("="*60)
        print("🚀 LOADING 3-MODEL SMART ENSEMBLE")
        print("="*60)

        self.v21 = YOLO(v21_path)  # 75.5% - Best overall
        self.v24 = YOLO(v24_path)  # 70.9% - Vegetable curries
        self.v25 = YOLO(v25_path)  # 70.5% - Specific foods!

        # Specific foods that V25 can detect (lowercase for matching)
        self.specific_foods = [
            'donut', 'eclair', 'cake', 'brownie', 'pastry',  # Sweets
            'cutlets', 'roll',  # Fried Filled
            'cream bun', 'crocodile bun', 'fish bun',  # Baked items
            'sausage hotdog'
        ]

        # Generic categories that should be REPLACED by V25 specific foods
        self.generic_categories = [
            'sweets', 'baked sweet bun', 'baked filled', 'fried filled'
        ]

        print("\n✅ All 3 models loaded successfully!")
        print(f"   V21: 75.5% mAP50 - Best overall")
        print(f"   V24: 70.9% mAP50 - Vegetable curries")
        print(f"   V25: 70.5% mAP50 - Specific sweets & snacks")

    def detect(self, image_path, confidence=0.25, verbose=True):
        """
        Smart detection: ONLY use V25 for specific foods,
        ignore generic categories when V25 detects something specific
        """

        # Step 1: Get V25 results (specific foods)
        v25_results = self.v25(image_path, conf=confidence)[0]
        v25_foods = [v25_results.names[int(box.cls[0])].lower() for box in v25_results.boxes]

        # Step 2: Get V21+V24 results for non-specific foods
        v21_results = self.v21(image_path, conf=confidence)[0]
        v24_results = self.v24(image_path, conf=confidence)[0]

        v21_foods = [v21_results.names[int(box.cls[0])].lower() for box in v21_results.boxes]
        v24_foods = [v24_results.names[int(box.cls[0])].lower() for box in v24_results.boxes]

        # Step 3: Check if V25 detected any specific foods
        specific_detected = [f for f in v25_foods if f in self.specific_foods]

        # Step 4: Collect other foods (non-generic, non-specific)
        other_foods = []

        for model_foods in [v21_foods, v24_foods]:
            for food in model_foods:
                # Skip generic categories that V25 is handling
                if food in self.generic_categories:
                    continue
                # Skip specific foods that V25 already detected
                if food in self.specific_foods:
                    continue
                # Add other unique foods
                if food not in other_foods:
                    other_foods.append(food)

        # Step 5: Build final result
        if specific_detected:
            # Use ONLY V25 specific foods + other non-generic foods
            final_foods = list(set(specific_detected + other_foods))
            if verbose:
                print(f"🎯 V25 specific foods: {specific_detected}")
                if other_foods:
                    print(f"   Other foods: {other_foods}")
                print(f"   ✅ FINAL: {final_foods}")
        else:
            # No specific foods, use V21+V24 ensemble
            final_foods = []
            for model_foods in [v21_foods, v24_foods]:
                for food in model_foods:
                    if food not in self.generic_categories and food not in final_foods:
                        final_foods.append(food)
            if verbose:
                print(f"🔍 No specific foods detected")
                print(f"   Using V21+V24: {final_foods}")

        return final_foods

    def detect_with_details(self, image_path, confidence=0.25):
        """Get detailed detection results"""
        print("\n" + "="*60)
        print(f"🔍 ANALYZING: {os.path.basename(image_path)}")
        print("="*60)

        # Individual model results
        print("\n📊 INDIVIDUAL MODEL DETECTIONS:")
        print("-"*40)

        v21_results = self.v21(image_path, conf=confidence)[0]
        v21_foods = [v21_results.names[int(box.cls[0])] for box in v21_results.boxes]
        print(f"   V21: {v21_foods}")

        v24_results = self.v24(image_path, conf=confidence)[0]
        v24_foods = [v24_results.names[int(box.cls[0])] for box in v24_results.boxes]
        print(f"   V24: {v24_foods}")

        v25_results = self.v25(image_path, conf=confidence)[0]
        v25_foods = [v25_results.names[int(box.cls[0])] for box in v25_results.boxes]
        print(f"   V25: {v25_foods}")

        # Ensemble result
        print("\n🎯 SMART ENSEMBLE RESULT (V25 takes priority):")
        print("-"*40)
        final_foods = self.detect(image_path, confidence, verbose=True)

        return {
            'v21': v21_foods,
            'v24': v24_foods,
            'v25': v25_foods,
            'ensemble': final_foods
        }

    def show_detection(self, image_path, confidence=0.25):
        """Show image with detection overlays"""
        foods = self.detect(image_path, confidence, verbose=False)

        img = cv2.imread(image_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        plt.figure(figsize=(12, 8))
        plt.imshow(img)
        plt.axis('off')
        plt.title(f"Detected: {', '.join(foods) if foods else 'No foods'}")
        plt.show()

        return foods


# ============================================================
# CREATE ENSEMBLE DETECTOR
# ============================================================

print("\n" + "="*60)
print("📥 CREATING SMART ENSEMBLE DETECTOR")
print("="*60)

ensemble = ThreeModelEnsemble(
    model_paths['V21'],
    model_paths['V24'],
    model_paths['V25']
)

# ============================================================
# UPLOAD AND TEST IMAGES
# ============================================================

print("\n" + "="*60)
print("📤 UPLOAD TEST IMAGES")
print("="*60)
print("Please upload food images to test (select multiple files):")
uploaded = files.upload()

# Save uploaded files
!mkdir -p test_images
image_paths = []
for filename in uploaded.keys():
    safe_filename = filename.replace(" ", "_")
    save_path = f"test_images/{safe_filename}"
    with open(save_path, 'wb') as f:
        f.write(uploaded[filename])
    image_paths.append(save_path)

print(f"\n✅ Uploaded {len(image_paths)} images")

# ============================================================
# TEST EACH IMAGE
# ============================================================

all_results = []

print("\n" + "="*60)
print("🔬 RUNNING TESTS")
print("="*60)

for img_path in image_paths:
    result = ensemble.detect_with_details(img_path)
    all_results.append(result)
    ensemble.show_detection(img_path)

# ============================================================
# SUMMARY TABLE
# ============================================================

print("\n" + "="*60)
print("📊 TEST SUMMARY")
print("="*60)

print("\n| Image | V21 | V24 | V25 | FINAL |")
print("|-------|-----|-----|-----|-------|")

for i, (img_path, result) in enumerate(zip(image_paths, all_results)):
    img_name = os.path.basename(img_path)[:20]
    v21_str = str(result['v21'][:2]) if result['v21'] else '[]'
    v24_str = str(result['v24'][:2]) if result['v24'] else '[]'
    v25_str = str(result['v25'][:2]) if result['v25'] else '[]'
    ensemble_str = str(result['ensemble'][:2]) if result['ensemble'] else '[]'
    print(f"| {img_name} | {v21_str} | {v24_str} | {v25_str} | {ensemble_str} |")

# ============================================================
# DETAILED RESULTS
# ============================================================

print("\n" + "="*60)
print("📋 DETAILED RESULTS")
print("="*60)

for i, (img_path, result) in enumerate(zip(image_paths, all_results)):
    print(f"\n{i+1}. {os.path.basename(img_path)}")
    print(f"   V21: {result['v21']}")
    print(f"   V24: {result['v24']}")
    print(f"   V25: {result['v25']}")
    print(f"   ✅ FINAL: {result['ensemble']}")

# ============================================================
# SAVE RESULTS
# ============================================================

import json
import datetime

timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
results_file = f'ensemble_results_{timestamp}.json'

with open(results_file, 'w') as f:
    json.dump(all_results, f, indent=2, default=str)

print(f"\n✅ Results saved to {results_file}")

# Download results
from google.colab import files
files.download(results_file)

print("\n" + "="*60)
print("🎉 SMART ENSEMBLE TEST COMPLETE!")
print("="*60)
print("""
Three models working together:
- V21 (75.5%): Best overall detection
- V24 (70.9%): Excellent for vegetable curries
- V25 (70.5%): Specific sweets and snacks!

Priority Rules:
- If V25 detects a specific food → ONLY that specific food
- Generic categories (sweets, baked sweet bun, etc.) are IGNORED
- Other foods (rice, curry, etc.) are added from V21/V24
""")
```
