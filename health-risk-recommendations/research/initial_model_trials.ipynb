```python
import pandas as pd
import numpy as np
```

Synetuc user data set

```python
data = {
    "age": [45, 28, 60, 35],
    "gender": ["M", "F", "M", "F"],
    "height": [1.65, 1.58, 1.70, 1.62],   # meters
    "weight": [82, 55, 90, 68],          # kg
    "activity_level": ["low", "moderate", "low", "moderate"],
    "daily_calories": [2800, 1800, 3000, 2200],
    "sugar": [65, 30, 70, 40],           # grams
    "sodium": [2600, 1200, 2800, 1800]   # mg
}

df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

BMI Calculation

```python
df["BMI"] = df["weight"] / (df["height"] ** 2)
df["BMI"] = df["BMI"].round(2)

df[["weight", "height", "BMI"]]
```

Categorize BMI

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

df["BMI_Category"] = df["BMI"].apply(bmi_category)
df[["BMI", "BMI_Category"]]
```

Health Risk Preidction

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

load  the data set

```python
import pandas as pd

df = pd.read_excel("/content/synthetic_health_risk_dataset_500.xlsx")
df.head()
```

```python
label_encoders = {}

categorical_columns = ["gender", "diabetes_risk", "hypertension_risk", "obesity_risk"]

for col in categorical_columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
```

Select inputs

```python
X = df[
    ["age", "gender", "daily_calories", "sugar_g",
     "sodium_mg", "carbohydrate_g", "fat_g"]
]
```

Disease wise Model training

Type -2 Dibates

```python
y_diabetes = df["diabetes_risk"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y_diabetes, test_size=0.2, random_state=42
)

diabetes_model = RandomForestClassifier(n_estimators=100, random_state=42)
diabetes_model.fit(X_train, y_train)

y_pred = diabetes_model.predict(X_test)

print("Diabetes Risk Prediction Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

Hypertension Risk

```python
y_hypertension = df["hypertension_risk"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y_hypertension, test_size=0.2, random_state=42
)

hypertension_model = RandomForestClassifier(n_estimators=100, random_state=42)
hypertension_model.fit(X_train, y_train)

y_pred = hypertension_model.predict(X_test)

print("Hypertension Risk Prediction Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

Obesity

```python
y_obesity = df["obesity_risk"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y_obesity, test_size=0.2, random_state=42
)

obesity_model = RandomForestClassifier(n_estimators=100, random_state=42)
obesity_model.fit(X_train, y_train)

y_pred = obesity_model.predict(X_test)

print("Obesity Risk Prediction Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

Personalized reccomendations demo

```python
import pandas as pd

# 1. Setup Sample Users (Simulating the database)
test_users = [
    {"id": "U101", "name": "Aruni", "age": 45, "gender": "F", "cal": 2800, "sugar": 75, "sodium": 2600, "carb": 350},
    {"id": "U102", "name": "Kasun", "age": 28, "gender": "M", "cal": 2100, "sugar": 20, "sodium": 1400, "carb": 220},
    {"id": "U103", "name": "Perera", "age": 60, "gender": "M", "cal": 3100, "sugar": 90, "sodium": 3500, "carb": 400}
]

def generate_report(user):
    # Prepare data for prediction
    # (Assuming your models are already loaded as 'diabetes_model', etc.)
    input_data = pd.DataFrame([{
        "age": user["age"],
        "gender": label_encoders["gender"].transform([user["gender"]])[0],
        "daily_calories": user["cal"],
        "sugar_g": user["sugar"],
        "sodium_mg": user["sodium"],
        "carbohydrate_g": user["carb"],
        "fat_g": 80 # Placeholder for fat
    }])

    # Get Predictions
    d_risk = label_encoders["diabetes_risk"].inverse_transform(diabetes_model.predict(input_data))[0]
    h_risk = label_encoders["hypertension_risk"].inverse_transform(hypertension_model.predict(input_data))[0]
    o_risk = label_encoders["obesity_risk"].inverse_transform(obesity_model.predict(input_data))[0]

    # --- RECOMMENDATION ENGINE LOGIC ---
    recommendations = []

    # Sugar/Diabetes Logic
    if d_risk == "High" or user["sugar"] > 50:
        recommendations.append(" SUGAR ALERT: Your intake is significantly high. Replace refined white sugar with Kitul Jaggery or Stevia.")
        recommendations.append(" STAPLE SWAP: Transition from White Rice to Red Basmati or Kurakkan-based meals to manage glycemic spikes.")

    # Sodium/Hypertension Logic
    if h_risk == "High" or user["sodium"] > 2300:
        recommendations.append(" SODIUM REDUCTION: Limit 'Bites', processed sausages, and excessive salt in your Pol Sambol.")
        recommendations.append(" HERBAL BOOST: Use more lime, ginger, and garlic in curries to enhance flavor without adding salt.")

    # Obesity/Calorie Logic
    if o_risk == "High" or user["cal"] > 2500:
        recommendations.append(" PORTION CONTROL: Follow the 'Healthy Plate' rule—fill 50% of your plate with local Mellun and vegetables.")

    # If everything is fine
    if not recommendations:
        recommendations.append(" Excellent balance! Continue your current dietary habits and stay hydrated.")

    return d_risk, h_risk, o_risk, recommendations

# 2. Loop through all customers and print reports
print("="*60)
print("      NUTRI SCANNER: PERSONALIZED HEALTH DASHBOARD      ")
print("="*60)

for user in test_users:
    d, h, o, advice = generate_report(user)

    print(f"\n[USER ID: {user['id']} | NAME: {user['name']}]")
    print(f"Stats: {user['cal']} kcal, {user['sugar']}g Sugar, {user['sodium']}mg Sodium")
    print(f"Predicted Risks: Diabetes: {d} | Hypertension: {h} | Obesity: {o}")
    print("-" * 30)
    print("PERSONALIZED ADVICE:")
    for item in advice:
        print(f"  * {item}")
    print("\n" + "."*60)
```
